```python
def category_total(data):
    """Recursively calculate the total of nested transaction categories."""
    total = 0

    for item in data:
        if isinstance(item, dict):
            total += category_total(item)
        elif isinstance(item, (int, float)):
            total += item

    return total


transactions = {
    "deposits": {
        "salary": 25000,
        "bonus": 5000
    },
    "withdrawals": {
        "atm": 3000,
        "shopping": 2500
    }
}

result = category_total(transactions)

print("Account Reference:", "ACC-1055")
print("Category Total:", result)
```

Q56. Validate arguments with a wrapper for banking case 2; use sample reference ACC-1056 and explain the result.

```python
def process_transfer(account, amount):
    print(f"Transfer of ₹{amount} processed for {account}")


def validate_transfer(function):
    def wrapper(account, amount):
        if not isinstance(account, str) or not account:
            print("Invalid account reference")
            return

        if not isinstance(amount, (int, float)) or amount <= 0:
            print("Invalid transfer amount")
            return

        function(account, amount)

    return wrapper


process_transfer = validate_transfer(process_transfer)

process_transfer("ACC-1056", 2500)
```

```python
def with_retry(task, max_attempts):
    def execute(amount):
        for attempt in range(1, max_attempts + 1):
            try:
                return task(amount)
            except RuntimeError as error:
                print("Attempt", attempt, ":", error)

        return "Transfer could not be completed"

    return execute


def bank_transfer(amount):
    if amount > 6000:
        raise RuntimeError("Daily transfer limit exceeded")

    return "Transfer successful for ₹" + str(amount)


retry_transfer = with_retry(bank_transfer, 3)

print("Account Reference:", "ACC-1098")
print(retry_transfer(4500))
```

```python
def add_transaction(history, amount):
    return (*history, amount)


def calculate_total(history):
    return sum(history)


account = "ACC-1099"

previous_transactions = (800, 1200)
updated_transactions = add_transaction(previous_transactions, 2500)

print("Account Reference:", account)
print("Previous Transactions:", previous_transactions)
print("Previous Total: ₹", calculate_total(previous_transactions))

print("Updated Transactions:", updated_transactions)
print("Updated Total: ₹", calculate_total(updated_transactions))
```


Q100. Design an end-to-end transactions analytics pipeline for banking case 3; use sample reference ACC-1100 and explain the result.

```python
from functools import reduce

transactions = [
    {"type": "deposit", "amount": 3000, "status": "success"},
    {"type": "withdrawal", "amount": 1200, "status": "success"},
    {"type": "deposit", "amount": 2500, "status": "failed"},
    {"type": "deposit", "amount": 1800, "status": "success"},
    {"type": "withdrawal", "amount": 700, "status": "success"}
]


def successful_transactions(data):
    return list(filter(lambda x: x["status"] == "success", data))


def get_amounts(data):
    return list(map(lambda x: x["amount"], data))


def find_total(values):
    return reduce(lambda x, y: x + y, values, 0)


valid = successful_transactions(transactions)

deposits = get_amounts(
    filter(lambda x: x["type"] == "deposit", valid)
)

withdrawals = get_amounts(
    filter(lambda x: x["type"] == "withdrawal", valid)
)

deposit_total = find_total(deposits)
withdrawal_total = find_total(withdrawals)

all_amounts = get_amounts(valid)
average = find_total(all_amounts) / len(all_amounts)

print("Account Reference:", "ACC-1100")
print("Successful Transactions:", len(valid))
print("Deposit Total: ₹", deposit_total)
print("Withdrawal Total: ₹", withdrawal_total)
print("Average Transaction: ₹", average)
```
