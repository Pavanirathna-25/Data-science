Q37. Implement a custom iterator for portfolio for banking case 1; use sample reference ACC-1037 and explain the result.

```python
class Portfolio:
    def __init__(self, items):
        self.items = items
        self.index = 0
    def __iter__(self):
        self.index = 0
        return self
    def __next__(self):
        if self.index < len(self.items):
            item = self.items[self.index]
            self.index += 1
            return item
        raise StopIteration
portfolio = Portfolio([
    "ACC-1037",
    "ACC-1038",
    "ACC-1039"
])
print("Portfolio Reference:", "ACC-1037")
for account in portfolio:
    print("Account:", account)
```

Q44. Implement a repository abstraction for banking case 1; use sample reference ACC-1044 and explain the result.

```python
from abc import ABC, abstractmethod
class AccountRepository(ABC):
    @abstractmethod
    def save(self, account):
        pass
    @abstractmethod
    def find(self, reference):
        pass
class MemoryAccountRepository(AccountRepository):
    def __init__(self):
        self.accounts = {}
    def save(self, account):
        self.accounts[account.reference] = account
    def find(self, reference):
        return self.accounts.get(reference)
class Account:
    def __init__(self, reference, holder, balance):
        self.reference = reference
        self.holder = holder
        self.balance = balance
repository = MemoryAccountRepository()
account = Account("ACC-1044", "Pavani", 50000)
repository.save(account)
result = repository.find("ACC-1044")
print("Account Reference:", result.reference)
print("Account Holder:", result.holder)
print("Balance: ₹", result.balance)
```

Q95. Implement a unit-of-work transaction for banking case 3; use sample reference ACC-1095 and explain the result.

```python
class BankAccount:
    def __init__(self, account_no, balance):
        self.account_no = account_no
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            raise Exception("Insufficient balance")
class UnitOfWork:
    def __init__(self, account):
        self.account = account
        self.old_balance = account.balance
    def commit(self):
        print("Transaction committed")
    def rollback(self):
        self.account.balance = self.old_balance
        print("Transaction rolled back")
account = BankAccount("ACC-1095", 12000)
work = UnitOfWork(account)
try:
    account.withdraw(3000)
    account.deposit(1500)
    work.commit()
except Exception as e:
    print(e)
    work.rollback()
print("Account:", account.account_no)
print("Final Balance:", account.balance)
```
