Assignment 17: Palindrome Checker

```python
## Write a function named is_palindrome().
def is_palindrome(text):
    text = text.lower()
    text = text.replace(" ", "")

    reversed_text = text[::-1]

    if text == reversed_text:
        return True
    else:
        return False

text = input("Enter text: ")

result = is_palindrome(text)

if result:
    print("The text is a palindrome.")
else:
    print("The text is not a palindrome.")
```

Assignment 34: Prime Number Function

```python
## Write a function named is_prime().
def is_prime(number):
    if number < 2:
        return False
    for i in range(2, number):
        if number % i == 0:
            return False
    return True
print("Prime Numbers from 1 to 100:")
for i in range(1, 101):
    if is_prime(i):
        print(i, end=" ")
```

Assignment 35: Fibonacci Function

```python
## Write a function named generate_fibonacci().
def generate_fibonacci(terms):
    if terms <= 0:
        return "Please enter a positive number."
    fibonacci = []
    a = 0
    b = 1
    for i in range(terms):
        fibonacci.append(a)
        c = a + b
        a = b
        b = c
    return fibonacci
terms = int(input("Enter number of terms: "))
result = generate_fibonacci(terms)
print(result)
```

Assignment 67: Filter Prime Numbers


```python
## Accept a list of integers.
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True
numbers = [10, 3, 5, 8, 11, 1, 0, -7, 13, 20]
prime_numbers = list(filter(is_prime, numbers))
print(prime_numbers)
```

Assignment 81: Calculate Factorial Using reduce()


```python
## Write a function that calculates factorial using reduce().
def factorial(n):
    if n < 0:
        return "Factorial not possible"
    if n == 0:
        return 1
    numbers = range(1, n + 1)
    return reduce(lambda a, b: a * b, numbers)
num = int(input("Enter a number: "))
print(factorial(num))
```

Assignment 92: Closure-Based Bank Balance


```python
## Write a function named create_bank_account().
def create_bank_account():
    balance = 1000
    def deposit(amount):
        nonlocal balance
        balance += amount
        return balance
    def withdraw(amount):
        nonlocal balance
        if amount <= balance:
            balance -= amount
            return balance
        return "Insufficient balance"
    def check_balance():
        return balance
    return deposit, withdraw, check_balance
deposit, withdraw, check_balance = create_bank_account()
print(deposit(500))
print(withdraw(300))
print(check_balance())
print(withdraw(2000))
```

Assignment 93: Function Decorator


```python
## Create a decorator named execution_logger().
def execution_logger(func):
    def wrapper():
        print("Starting function...")
        result = func()
        print("Function completed.")
        return result
    return wrapper
@execution_logger
def addition():
    return 10 + 20
result = addition()
print("Result:", result)
```
