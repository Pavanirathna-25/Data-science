
Q33. Process transfers while isolating bad rows for banking case 1; use sample reference ACC-1033 and explain the result.

```python
account_ref = "ACC-1033"
balance = 15000
transfers = [
    ("TXN001", 3000),
    ("TXN002", "abc"),
    ("TXN003", 2000),
    ("TXN004", -500)
]
successful_transfers = []
bad_transfers = []
for transaction_id, amount_input in transfers:
    try:
        amount = float(amount_input)
        if amount <= 0:
            raise ValueError("Amount must be greater than zero")
        if amount > balance:
            raise ValueError("Insufficient balance")
        balance -= amount
        successful_transfers.append(transaction_id)
    except ValueError as e:
        bad_transfers.append((transaction_id, str(e)))
print("Account:", account_ref)
print("Successful transfers:", successful_transfers)
print("Bad transfers:", bad_transfers)
print("Final balance:", balance)
```

Q34. Retry a transient payment gateway for banking case 1; use sample reference ACC-1034 and explain the result.

```python
class PaymentGatewayError(Exception):
    pass
account_ref = "ACC-1034"
balance = 10000
payment = 3000
attempts = 0
max_retries = 3
while attempts < max_retries:
    try:
        attempts += 1
        print("Payment attempt:", attempts)
        if attempts < 3:
            raise PaymentGatewayError("Payment gateway temporarily unavailable")
        balance -= payment
        print("Payment successful")
        print("Remaining balance:", balance)
        break
    except PaymentGatewayError as e:
        print("Gateway error:", e)
        if attempts == max_retries:
            print("Payment failed after maximum retries")
```
